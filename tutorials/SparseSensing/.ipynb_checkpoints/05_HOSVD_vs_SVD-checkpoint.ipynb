# <center>Case Study 5: 2D HOSVD vs SVD</center>

Compare flat SVD and HOSVD sparse sensing on a 2D parameter-time tensor, run the HOSVD case through RAVEN, and animate the reconstructed 2D field in time.

```python
import os
import sys
import textwrap
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML
import numpy as np
import pandas as pd
import pysensors as ps


def find_repo_root(start=Path.cwd()):
    for path in (start, *start.parents):
        if (path / "ravenframework").is_dir() and (path / "tests").is_dir():
            return path
    raise RuntimeError("Could not locate the repository root from the current working directory.")


repo_root = find_repo_root()
plt.rcParams.update({"figure.dpi": 120, "axes.grid": True, "grid.alpha": 0.22})
svd_random_seed = 0

import importlib.util
module_path = repo_root / "ravenframework" / "Models" / "PostProcessors" / "SparseSensingBases.py"
spec = importlib.util.spec_from_file_location("SparseSensingBases", module_path)
SparseSensingBases = importlib.util.module_from_spec(spec)
spec.loader.exec_module(SparseSensingBases)
HOSVDBasis = SparseSensingBases.HOSVDBasis
```

## Build a 2D parameter-time tensor

```python
NX = 18
NY = 12
NT = 8
x_axis = np.linspace(0.0, 1.0, NX)
y_axis = np.linspace(0.0, 1.0, NY)
times = np.linspace(0.0, 1.0, NT)
XX, YY = np.meshgrid(x_axis, y_axis, indexing="xy")
X = XX.ravel()
Y = YY.ravel()
point_id = np.arange(X.size)


def two_dimensional_field(alpha):
    left = np.exp(-(((X - (0.25 + 0.10 * alpha)) / 0.16) ** 2 + ((Y - 0.35) / 0.18) ** 2))
    right = np.exp(-(((X - 0.72) / 0.18) ** 2 + ((Y - (0.68 - 0.08 * alpha)) / 0.16) ** 2))
    wave = np.sin((2.0 + alpha) * np.pi * X) * np.cos(2.0 * np.pi * Y)
    return (
        np.exp(-0.85 * times[:, None]) * ((2.0 + 0.25 * alpha) * left[None, :] + (0.6 + 1.35 * alpha) * right[None, :])
        + 0.28 * np.cos(2.0 * np.pi * times[:, None]) * wave[None, :]
    )


def field_image(values):
    return np.asarray(values, dtype=float).reshape(NY, NX)


def spatial_frame(ax, values, title, limits, cmap="coolwarm", sensors=None, schedule_times=None, frame_time=None):
    image = ax.imshow(
        field_image(values),
        extent=[x_axis.min(), x_axis.max(), y_axis.min(), y_axis.max()],
        origin="lower",
        cmap=cmap,
        vmin=limits[0],
        vmax=limits[1],
        interpolation="bicubic",
        aspect="equal",
    )
    if sensors is not None:
        sensors = np.asarray(sensors, dtype=int)
        ax.scatter(X[sensors], Y[sensors], facecolors="none", edgecolors="black", s=88, linewidths=1.2)
    if schedule_times is not None and frame_time is not None:
        schedule_times = np.asarray(schedule_times)
        active = np.isclose(schedule_times, frame_time)
        if np.any(active):
            active_points = np.asarray(sensors, dtype=int)[active]
            ax.scatter(X[active_points], Y[active_points], marker="x", c="black", s=92, linewidths=1.8)
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_aspect("equal", adjustable="box")
    return image


def style_sensor_comparison(table, caption, columns=None):
    if columns is None:
        columns = [
            "match", "order_match", "sensor_id", "pySensors order", "RAVEN order",
            "pySensors value", "RAVEN value", "pySensors x", "RAVEN x", "pySensors y", "RAVEN y",
        ]

    def highlight(row):
        if bool(row.get("match", False)) and bool(row.get("order_match", True)):
            return ["background-color: #d9ead3; color: #073b18" for _ in row]
        if bool(row.get("match", False)):
            return ["background-color: #fff2cc; color: #4a3600" for _ in row]
        return ["background-color: #f4cccc; color: #5f1010" for _ in row]

    fmt = {
        "match": lambda value: "yes" if value else "no",
        "order_match": lambda value: "yes" if value else "no",
        "pySensors order": "{:.0f}",
        "RAVEN order": "{:.0f}",
        "pySensors value": "{:.6f}",
        "RAVEN value": "{:.6f}",
        "pySensors x": "{:.6f}",
        "RAVEN x": "{:.6f}",
        "pySensors y": "{:.6f}",
        "RAVEN y": "{:.6f}",
        "pySensors time": "{:.6f}",
        "RAVEN time": "{:.6f}",
    }
    return table[columns].style.apply(highlight, axis=1).format(fmt, na_rep="-").hide(axis="index").set_caption(caption)


def nearest_point_indices(frame):
    grid = np.column_stack([X, Y])
    indices = []
    distances = []
    for _, row in frame.iterrows():
        point = np.array([row["x"], row["y"]])
        distance = np.linalg.norm(grid - point, axis=1)
        index = int(np.argmin(distance))
        indices.append(index)
        distances.append(float(distance[index]))
    return np.asarray(indices, dtype=int), np.asarray(distances)

alphas = np.linspace(0.1, 1.0, 5)
tensor = np.stack([two_dimensional_field(alpha) for alpha in alphas])
extra_mode = (
    0.18
    * np.linspace(-1.0, 1.0, len(alphas))[:, None, None]
    * np.sin(np.pi * times)[None, :, None]
    * np.sin(3.0 * np.pi * X)[None, None, :]
)
T = tensor + extra_mode
snapshot_matrix = T.reshape(len(alphas) * len(times), X.size)
print(f"tensor shape: {T.shape}")
print(f"snapshot matrix shape: {snapshot_matrix.shape}")
```

```python
ranks = range(1, 13)
svd_errors = []
hosvd_errors = []
for rank in ranks:
    svd_basis_model = ps.SSPOR(basis=ps.basis.SVD(n_basis_modes=rank, random_state=42), n_sensors=rank, optimizer=ps.optimizers.QR())
    svd_basis_model.fit(snapshot_matrix, seed=42)
    svd_s = svd_basis_model.get_selected_sensors()
    svd_hat = svd_basis_model.predict(snapshot_matrix[:, svd_s], method="unregularized")
    svd_errors.append(np.linalg.norm(snapshot_matrix - svd_hat) / np.linalg.norm(snapshot_matrix))

    hosvd_basis = HOSVDBasis(n_basis_modes=rank).fit(T)
    hosvd_basis_model = ps.SSPOR(basis=hosvd_basis, n_sensors=rank, optimizer=ps.optimizers.QR())
    hosvd_basis_model.fit(snapshot_matrix, prefit_basis=True, seed=42)
    hosvd_s = hosvd_basis_model.get_selected_sensors()
    hosvd_hat = hosvd_basis_model.predict(snapshot_matrix[:, hosvd_s], method="unregularized")
    hosvd_errors.append(np.linalg.norm(snapshot_matrix - hosvd_hat) / np.linalg.norm(snapshot_matrix))
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(list(ranks), svd_errors, "o-", label="flat SVD")
ax.plot(list(ranks), hosvd_errors, "s--", label="HOSVD spatial factor")
ax.set_xlabel("basis rank")
ax.set_ylabel("unregularized relative L2 error")
ax.set_title("SVD vs HOSVD rank sweep")
ax.legend()
plt.show()
```

## pySensors HOSVD reconstruction

```python
n_modes = 12
n_sensors = 12
svd_model = ps.SSPOR(basis=ps.basis.SVD(n_basis_modes=n_modes, random_state=42), n_sensors=n_sensors, optimizer=ps.optimizers.QR())
svd_model.fit(snapshot_matrix, seed=42)
svd_sensors = svd_model.get_selected_sensors()
svd_regularized_reconstruction = svd_model.predict(snapshot_matrix[:, svd_sensors])
svd_regularized_error = np.linalg.norm(snapshot_matrix - svd_regularized_reconstruction) / np.linalg.norm(snapshot_matrix)
svd_reconstruction = svd_model.predict(snapshot_matrix[:, svd_sensors], method="unregularized")
svd_error = np.linalg.norm(snapshot_matrix - svd_reconstruction) / np.linalg.norm(snapshot_matrix)

hosvd_basis = HOSVDBasis(n_basis_modes=n_modes).fit(T)
hosvd_model = ps.SSPOR(basis=hosvd_basis, n_sensors=n_sensors, optimizer=ps.optimizers.QR())
hosvd_model.fit(snapshot_matrix, prefit_basis=True, seed=42)
hosvd_sensors = hosvd_model.get_selected_sensors()
hosvd_regularized_reconstruction = hosvd_model.predict(snapshot_matrix[:, hosvd_sensors])
hosvd_regularized_error = np.linalg.norm(snapshot_matrix - hosvd_regularized_reconstruction) / np.linalg.norm(snapshot_matrix)
hosvd_reconstruction = hosvd_model.predict(snapshot_matrix[:, hosvd_sensors], method="unregularized")
hosvd_error = np.linalg.norm(snapshot_matrix - hosvd_reconstruction) / np.linalg.norm(snapshot_matrix)

pd.DataFrame({"method": ["flat SVD", "HOSVD spatial factor"], "sensor_ids": [list(svd_sensors), list(hosvd_sensors)], "regularized_default_error": [svd_regularized_error, hosvd_regularized_error], "unregularized_relative_l2_error": [svd_error, hosvd_error]})
```

## Run the HOSVD selection through RAVEN

```python
raven_case_dir = repo_root / "tutorials" / "SparseSensing" / "_raven_runs" / "hosvd_vs_svd"
raven_case_dir.mkdir(parents=True, exist_ok=True)
raven_alpha_values = " ".join(f"{value:.17g}" for value in alphas)

(raven_case_dir / "genNotebookHOSVD2D.py").write_text(
    textwrap.dedent(
        """\

        import numpy as np

        NX = 18
        NY = 12
        NT = 8
        X_AXIS = np.linspace(0.0, 1.0, NX)
        Y_AXIS = np.linspace(0.0, 1.0, NY)
        TIMES = np.linspace(0.0, 1.0, NT)
        XX, YY = np.meshgrid(X_AXIS, Y_AXIS, indexing='xy')
        X = XX.ravel()
        Y = YY.ravel()

        def read_alpha(Input, default=0.5):
          if isinstance(Input, dict):
            raw = Input.get('alpha', default)
          else:
            raw = getattr(Input, 'alpha', default)
          return float(np.asarray(raw).reshape(-1)[0])

        def two_dimensional_field(alpha):
          left = np.exp(-(((X - (0.25 + 0.10 * alpha)) / 0.16) ** 2 + ((Y - 0.35) / 0.18) ** 2))
          right = np.exp(-(((X - 0.72) / 0.18) ** 2 + ((Y - (0.68 - 0.08 * alpha)) / 0.16) ** 2))
          wave = np.sin((2.0 + alpha) * np.pi * X) * np.cos(2.0 * np.pi * Y)
          return (
              np.exp(-0.85 * TIMES[:, None]) * ((2.0 + 0.25 * alpha) * left[None, :] + (0.6 + 1.35 * alpha) * right[None, :])
              + 0.28 * np.cos(2.0 * np.pi * TIMES[:, None]) * wave[None, :]
          )

        ALPHAS = np.linspace(0.1, 1.0, 5)

        def full_tensor():
          tensor = np.stack([two_dimensional_field(alpha) for alpha in ALPHAS])
          extra_mode = (
              0.18
              * np.linspace(-1.0, 1.0, len(ALPHAS))[:, None, None]
              * np.sin(np.pi * TIMES)[None, :, None]
              * np.sin(3.0 * np.pi * X)[None, None, :]
          )
          return tensor + extra_mode

        def run(self, Input):
          alpha = read_alpha(Input)
          alpha_index = int(np.argmin(np.abs(ALPHAS - alpha)))
          self.time = TIMES
          self.pointID = np.arange(X.size)
          self.x = X.copy()
          self.y = Y.copy()
          self.T = full_tensor()[alpha_index]
          self._indexMap = {'x': ['pointID'], 'y': ['pointID'], 'T': ['time', 'pointID']}
        """
    )
)

raven_input = raven_case_dir / "hosvd_vs_svd.xml"
raven_input.write_text(
    textwrap.dedent(
        f"""\
        <?xml version="1.0" ?>
        <Simulation verbosity="quiet">
          <RunInfo><WorkingDir>.</WorkingDir><Sequence>sample, pp, print</Sequence><batchSize>1</batchSize></RunInfo>
          <Models><ExternalModel name="genT" subType="" ModuleToLoad="genNotebookHOSVD2D"><inputs>alpha</inputs><outputs>time, pointID, x, y, T</outputs></ExternalModel><PostProcessor name="mySPSL" subType="SparseSensing"><Goal subType="reconstruction"><features>x, y, T</features><target>T</target><basis>HOSVD</basis><nModes>{n_modes}</nModes><nSensors>{n_sensors}</nSensors><optimizer>QR</optimizer><pivotParameter>time</pivotParameter><reshape>snapshot</reshape><reconstructionMethod>auto</reconstructionMethod><seed>42</seed></Goal></PostProcessor></Models>
          <Distributions><Uniform name="uniAlpha"><lowerBound>0.0</lowerBound><upperBound>1.0</upperBound></Uniform></Distributions>
          <Samplers><Grid name="alphaGrid"><variable name="alpha"><distribution>uniAlpha</distribution><grid construction="custom" type="value">{raven_alpha_values}</grid></variable></Grid></Samplers>
          <Steps><MultiRun name="sample"><Input class="DataObjects" type="PointSet">placeholder</Input><Model class="Models" type="ExternalModel">genT</Model><Sampler class="Samplers" type="Grid">alphaGrid</Sampler><Output class="DataObjects" type="DataSet">dataDS</Output></MultiRun><PostProcess name="pp"><Input class="DataObjects" type="DataSet">dataDS</Input><Model class="Models" type="PostProcessor">mySPSL</Model><Output class="DataObjects" type="DataSet">outPP</Output></PostProcess><IOStep name="print"><Input class="DataObjects" type="DataSet">outPP</Input><Output class="OutStreams" type="Print">outPP</Output></IOStep></Steps>
          <DataObjects><PointSet name="placeholder"><Input>alpha</Input><Output>OutputPlaceHolder</Output></PointSet><DataSet name="dataDS"><Input>alpha</Input><Output>x, y, T</Output><Index var="time">T</Index><Index var="pointID">x, y, T</Index></DataSet><DataSet name="outPP"><Output>x, y, T</Output><Index var="sensor">x, y, T</Index></DataSet></DataObjects>
          <OutStreams><Print name="outPP"><type>csv</type><source>outPP</source><what>output</what></Print></OutStreams>
        </Simulation>
        """
    )
)
print(f"RAVEN input: {raven_input}")
```

```python
raven_case_dir.mkdir(parents=True, exist_ok=True)
raven_output = raven_case_dir / "outPP.csv"
if raven_output.exists():
    raven_output.unlink()

import subprocess
run_env = os.environ.copy()
run_env["PATH"] = f"{repo_root}{os.pathsep}{run_env.get('PATH', '')}"
run_env["INSTALLATION_MANAGER"] = "CONDA"
run_env["RAVEN_LIBS_NAME"] = "raven_spsl_043"
run_env["PYTHON_COMMAND"] = sys.executable
run_env.setdefault("MPLCONFIGDIR", str(Path("/tmp") / "mpl_raven"))

print(f"$ raven_framework {raven_input.name}")
run_result = subprocess.run(
    ["raven_framework", raven_input.name],
    cwd=str(raven_case_dir),
    env=run_env,
    text=True,
    capture_output=True,
)
print(run_result.stdout)
if run_result.stderr:
    print(run_result.stderr)
if run_result.returncode != 0:
    raise RuntimeError(f"RAVEN failed with exit code {run_result.returncode}")
if not raven_output.exists():
    raise RuntimeError(f"RAVEN completed but did not write {raven_output}")
print(f"Fresh RAVEN run completed: {raven_output}")

raven_sensor_table = pd.read_csv(raven_output)
raven_indices, raven_distances = nearest_point_indices(raven_sensor_table)
pysensors_compare = pd.DataFrame({"sensor_id": hosvd_sensors.astype(int), "pySensors order": np.arange(1, len(hosvd_sensors) + 1), "pySensors value": snapshot_matrix[0, hosvd_sensors], "pySensors x": X[hosvd_sensors], "pySensors y": Y[hosvd_sensors]})
raven_compare = pd.DataFrame({"sensor_id": raven_indices, "RAVEN order": raven_sensor_table["sensor"].to_numpy(), "RAVEN value": raven_sensor_table["T"].to_numpy(), "RAVEN x": raven_sensor_table["x"].to_numpy(), "RAVEN y": raven_sensor_table["y"].to_numpy(), "nearest_grid_distance": raven_distances})
comparison_table = pysensors_compare.merge(raven_compare, on="sensor_id", how="outer").sort_values(["pySensors order", "RAVEN order"], na_position="last").reset_index(drop=True)
comparison_table["same sensor"] = comparison_table["pySensors order"].notna() & comparison_table["RAVEN order"].notna()
comparison_table["same location"] = np.isclose(comparison_table["pySensors x"], comparison_table["RAVEN x"]) & np.isclose(comparison_table["pySensors y"], comparison_table["RAVEN y"])
comparison_table["same value"] = np.isclose(comparison_table["pySensors value"], comparison_table["RAVEN value"])
comparison_table["match"] = comparison_table[["same sensor", "same location", "same value"]].all(axis=1)
comparison_table["order_match"] = comparison_table["pySensors order"].eq(comparison_table["RAVEN order"])
print(f"same selected HOSVD sensor set: {comparison_table['match'].all() and len(comparison_table) == n_sensors}")
print(f"same selected HOSVD sensor order: {np.array_equal(hosvd_sensors, raven_indices)}")
style_sensor_comparison(comparison_table, "RAVEN vs pySensors HOSVD sensors")
```

## 2D HOSVD reconstruction animation

```python
hosvd_truth_tensor = snapshot_matrix.reshape(len(alphas), len(times), X.size)
hosvd_reconstruction_tensor = hosvd_reconstruction.reshape(len(alphas), len(times), X.size)
animation_alpha_index = 2
truth_animation_matrix = hosvd_truth_tensor[animation_alpha_index]
reconstruction_animation_matrix = hosvd_reconstruction_tensor[animation_alpha_index]
relative_error_animation_matrix = (reconstruction_animation_matrix - truth_animation_matrix) / max(np.max(np.abs(truth_animation_matrix)), 1e-12)
time_relative_l2_error = np.linalg.norm(reconstruction_animation_matrix - truth_animation_matrix, axis=1) / np.maximum(np.linalg.norm(truth_animation_matrix, axis=1), 1e-12)
field_limits = (truth_animation_matrix.min(), truth_animation_matrix.max())
error_limit = max(np.max(np.abs(relative_error_animation_matrix)), 1e-12)
fig, axes = plt.subplots(1, 3, figsize=(13.5, 4.2), sharex=True, sharey=True)
truth_sc = spatial_frame(axes[0], truth_animation_matrix[0], "Truth", field_limits, sensors=hosvd_sensors)
recon_sc = spatial_frame(axes[1], reconstruction_animation_matrix[0], "Reconstruction", field_limits, sensors=hosvd_sensors)
err_sc = spatial_frame(axes[2], relative_error_animation_matrix[0], "Relative error", (-error_limit, error_limit), cmap="seismic", sensors=hosvd_sensors)
for ax in axes:
    ax.set_ylabel("y")
fig.colorbar(truth_sc, ax=axes[0], shrink=0.8, label="temperature")
fig.colorbar(recon_sc, ax=axes[1], shrink=0.8, label="temperature")
fig.colorbar(err_sc, ax=axes[2], shrink=0.8, label="relative error")

def update(frame):
    truth_sc.set_data(field_image(truth_animation_matrix[frame]))
    recon_sc.set_data(field_image(reconstruction_animation_matrix[frame]))
    err_sc.set_data(field_image(relative_error_animation_matrix[frame]))
    fig.suptitle(f"HOSVD alpha={alphas[animation_alpha_index]:.3f}, time={times[frame]:.3f}; rel L2={time_relative_l2_error[frame]:.3e}")
    return truth_sc, recon_sc, err_sc

hosvd_field_animation = animation.FuncAnimation(fig, update, frames=len(times), interval=650, blit=False)
plt.close(fig)
HTML(hosvd_field_animation.to_jshtml())
```
