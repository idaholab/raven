# <center>Case Study 4: 2D Spatiotemporal Sensor Schedule</center>

Select scheduled `(x, y, time)` measurements with pySensors, replicate the schedule with RAVEN, and animate the held-out 2D reconstruction through time.

```python
import os
import sys
import textwrap
from pathlib import Path

import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML
import numpy as np
import pandas as pd
import pysensors as ps


def find_repo_root(start=Path.cwd()):
    for path in (start, *start.parents):
        if (path / "ravenframework").is_dir() and (path / "tests").is_dir():
            return path
    raise RuntimeError("Could not locate the repository root from the current working directory.")


repo_root = find_repo_root()
plt.rcParams.update({"figure.dpi": 120, "axes.grid": True, "grid.alpha": 0.22})
svd_random_seed = 0
```

## Build a 2D spatiotemporal schedule problem

```python
NX = 18
NY = 12
NT = 8
x_axis = np.linspace(0.0, 1.0, NX)
y_axis = np.linspace(0.0, 1.0, NY)
times = np.linspace(0.0, 1.0, NT)
XX, YY = np.meshgrid(x_axis, y_axis, indexing="xy")
X = XX.ravel()
Y = YY.ravel()
point_id = np.arange(X.size)


def two_dimensional_field(alpha):
    left = np.exp(-(((X - (0.25 + 0.10 * alpha)) / 0.16) ** 2 + ((Y - 0.35) / 0.18) ** 2))
    right = np.exp(-(((X - 0.72) / 0.18) ** 2 + ((Y - (0.68 - 0.08 * alpha)) / 0.16) ** 2))
    wave = np.sin((2.0 + alpha) * np.pi * X) * np.cos(2.0 * np.pi * Y)
    return (
        np.exp(-0.85 * times[:, None]) * ((2.0 + 0.25 * alpha) * left[None, :] + (0.6 + 1.35 * alpha) * right[None, :])
        + 0.28 * np.cos(2.0 * np.pi * times[:, None]) * wave[None, :]
    )


def field_image(values):
    return np.asarray(values, dtype=float).reshape(NY, NX)


def spatial_frame(ax, values, title, limits, cmap="coolwarm", sensors=None, schedule_times=None, frame_time=None):
    image = ax.imshow(
        field_image(values),
        extent=[x_axis.min(), x_axis.max(), y_axis.min(), y_axis.max()],
        origin="lower",
        cmap=cmap,
        vmin=limits[0],
        vmax=limits[1],
        interpolation="bicubic",
        aspect="equal",
    )
    if sensors is not None:
        sensors = np.asarray(sensors, dtype=int)
        ax.scatter(X[sensors], Y[sensors], facecolors="none", edgecolors="black", s=88, linewidths=1.2)
    if schedule_times is not None and frame_time is not None:
        schedule_times = np.asarray(schedule_times)
        active = np.isclose(schedule_times, frame_time)
        if np.any(active):
            active_points = np.asarray(sensors, dtype=int)[active]
            ax.scatter(X[active_points], Y[active_points], marker="x", c="black", s=92, linewidths=1.8)
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_aspect("equal", adjustable="box")
    return image


def style_sensor_comparison(table, caption, columns=None):
    if columns is None:
        columns = [
            "match", "order_match", "sensor_id", "pySensors order", "RAVEN order",
            "pySensors value", "RAVEN value", "pySensors x", "RAVEN x", "pySensors y", "RAVEN y",
        ]

    def highlight(row):
        if bool(row.get("match", False)) and bool(row.get("order_match", True)):
            return ["background-color: #d9ead3; color: #073b18" for _ in row]
        if bool(row.get("match", False)):
            return ["background-color: #fff2cc; color: #4a3600" for _ in row]
        return ["background-color: #f4cccc; color: #5f1010" for _ in row]

    fmt = {
        "match": lambda value: "yes" if value else "no",
        "order_match": lambda value: "yes" if value else "no",
        "pySensors order": "{:.0f}",
        "RAVEN order": "{:.0f}",
        "pySensors value": "{:.6f}",
        "RAVEN value": "{:.6f}",
        "pySensors x": "{:.6f}",
        "RAVEN x": "{:.6f}",
        "pySensors y": "{:.6f}",
        "RAVEN y": "{:.6f}",
        "pySensors time": "{:.6f}",
        "RAVEN time": "{:.6f}",
    }
    return table[columns].style.apply(highlight, axis=1).format(fmt, na_rep="-").hide(axis="index").set_caption(caption)


def nearest_point_indices(frame):
    grid = np.column_stack([X, Y])
    indices = []
    distances = []
    for _, row in frame.iterrows():
        point = np.array([row["x"], row["y"]])
        distance = np.linalg.norm(grid - point, axis=1)
        index = int(np.argmin(distance))
        indices.append(index)
        distances.append(float(distance[index]))
    return np.asarray(indices, dtype=int), np.asarray(distances)

alphas = np.linspace(0.1, 1.0, 5)
T = np.stack([two_dimensional_field(alpha) for alpha in alphas])
schedule_matrix = T.transpose(0, 2, 1).reshape(len(alphas), X.size * len(times))
print(f"tensor shape: {T.shape}")
print(f"schedule matrix shape: {schedule_matrix.shape}")
```

```python
n_sensors = 8
model = ps.SSPOR(basis=ps.basis.Identity(n_basis_modes=len(alphas)), n_sensors=n_sensors, optimizer=ps.optimizers.QR())
model.fit(schedule_matrix, seed=42)
selected = model.get_selected_sensors()
space_idx = selected // len(times)
time_idx = selected % len(times)
schedule_table = pd.DataFrame({"sensor_order": np.arange(1, len(selected) + 1), "column_id": selected, "point_id": space_idx, "x": X[space_idx], "y": Y[space_idx], "time": times[time_idx], "first_alpha_value": schedule_matrix[0, selected]})
schedule_table
```

## Run the same schedule selection through RAVEN

```python
raven_case_dir = repo_root / "tutorials" / "SparseSensing" / "_raven_runs" / "spatiotemporal_schedule"
raven_case_dir.mkdir(parents=True, exist_ok=True)
raven_alpha_values = " ".join(f"{value:.17g}" for value in alphas)

(raven_case_dir / "genNotebookSpatiotemporal2D.py").write_text(
    textwrap.dedent(
        """\

        import numpy as np

        NX = 18
        NY = 12
        NT = 8
        X_AXIS = np.linspace(0.0, 1.0, NX)
        Y_AXIS = np.linspace(0.0, 1.0, NY)
        TIMES = np.linspace(0.0, 1.0, NT)
        XX, YY = np.meshgrid(X_AXIS, Y_AXIS, indexing='xy')
        X = XX.ravel()
        Y = YY.ravel()

        def read_alpha(Input, default=0.5):
          if isinstance(Input, dict):
            raw = Input.get('alpha', default)
          else:
            raw = getattr(Input, 'alpha', default)
          return float(np.asarray(raw).reshape(-1)[0])

        def two_dimensional_field(alpha):
          left = np.exp(-(((X - (0.25 + 0.10 * alpha)) / 0.16) ** 2 + ((Y - 0.35) / 0.18) ** 2))
          right = np.exp(-(((X - 0.72) / 0.18) ** 2 + ((Y - (0.68 - 0.08 * alpha)) / 0.16) ** 2))
          wave = np.sin((2.0 + alpha) * np.pi * X) * np.cos(2.0 * np.pi * Y)
          return (
              np.exp(-0.85 * TIMES[:, None]) * ((2.0 + 0.25 * alpha) * left[None, :] + (0.6 + 1.35 * alpha) * right[None, :])
              + 0.28 * np.cos(2.0 * np.pi * TIMES[:, None]) * wave[None, :]
          )


        def run(self, Input):
          alpha = read_alpha(Input)
          self.time = TIMES
          self.pointID = np.arange(X.size)
          self.x = X.copy()
          self.y = Y.copy()
          self.T = two_dimensional_field(alpha)
          self._indexMap = {'x': ['pointID'], 'y': ['pointID'], 'T': ['time', 'pointID']}
        """
    )
)

raven_input = raven_case_dir / "spatiotemporal_schedule.xml"
raven_input.write_text(
    textwrap.dedent(
        f"""\
        <?xml version="1.0" ?>
        <Simulation verbosity="quiet">
          <RunInfo><WorkingDir>.</WorkingDir><Sequence>sample, pp, print</Sequence><batchSize>1</batchSize></RunInfo>
          <Models><ExternalModel name="genT" subType="" ModuleToLoad="genNotebookSpatiotemporal2D"><inputs>alpha</inputs><outputs>time, pointID, x, y, T</outputs></ExternalModel><PostProcessor name="mySPSL" subType="SparseSensing"><Goal subType="reconstruction"><features>x, y, T</features><target>T</target><basis>Identity</basis><nModes>{len(alphas)}</nModes><nSensors>{n_sensors}</nSensors><optimizer>QR</optimizer><pivotParameter>time</pivotParameter><reshape>spatiotemporal</reshape><reconstructionMethod>auto</reconstructionMethod><seed>42</seed></Goal></PostProcessor></Models>
          <Distributions><Uniform name="uniAlpha"><lowerBound>0.0</lowerBound><upperBound>1.0</upperBound></Uniform></Distributions>
          <Samplers><Grid name="alphaGrid"><variable name="alpha"><distribution>uniAlpha</distribution><grid construction="custom" type="value">{raven_alpha_values}</grid></variable></Grid></Samplers>
          <Steps><MultiRun name="sample"><Input class="DataObjects" type="PointSet">placeholder</Input><Model class="Models" type="ExternalModel">genT</Model><Sampler class="Samplers" type="Grid">alphaGrid</Sampler><Output class="DataObjects" type="DataSet">dataDS</Output></MultiRun><PostProcess name="pp"><Input class="DataObjects" type="DataSet">dataDS</Input><Model class="Models" type="PostProcessor">mySPSL</Model><Output class="DataObjects" type="DataSet">outPP</Output></PostProcess><IOStep name="print"><Input class="DataObjects" type="DataSet">outPP</Input><Output class="OutStreams" type="Print">outPP</Output></IOStep></Steps>
          <DataObjects><PointSet name="placeholder"><Input>alpha</Input><Output>OutputPlaceHolder</Output></PointSet><DataSet name="dataDS"><Input>alpha</Input><Output>x, y, T</Output><Index var="time">T</Index><Index var="pointID">x, y, T</Index></DataSet><DataSet name="outPP"><Output>x, y, T, time</Output><Index var="sensor">x, y, T, time</Index></DataSet></DataObjects>
          <OutStreams><Print name="outPP"><type>csv</type><source>outPP</source><what>output</what></Print></OutStreams>
        </Simulation>
        """
    )
)
print(f"RAVEN input: {raven_input}")
```

```python
raven_output = raven_case_dir / "outPP.csv"
if raven_output.exists():
    raven_output.unlink()

notebook_cwd = Path.cwd()
os.environ["PATH"] = f"{repo_root}{os.pathsep}{os.environ.get('PATH', '')}"
os.environ["INSTALLATION_MANAGER"] = "CONDA"
os.environ["RAVEN_LIBS_NAME"] = "raven_spsl_043"
os.environ["PYTHON_COMMAND"] = sys.executable
os.environ.setdefault("MPLCONFIGDIR", str(Path("/tmp") / "mpl_raven"))

%cd {raven_case_dir}
!raven_framework {raven_input.name}
%cd {notebook_cwd}

if not raven_output.exists():
    raise RuntimeError(f"RAVEN completed but did not write {raven_output}")
print(f"Fresh RAVEN run completed: {raven_output}")

raven_sensor_table = pd.read_csv(raven_output)
raven_space_idx, raven_distances = nearest_point_indices(raven_sensor_table)
raven_time_idx = np.array([int(np.argmin(np.abs(times - row["time"]))) for _, row in raven_sensor_table.iterrows()])
raven_column_id = raven_space_idx * len(times) + raven_time_idx
pysensors_compare = pd.DataFrame({"column_id": selected.astype(int), "sensor_id": space_idx.astype(int), "pySensors order": np.arange(1, len(selected) + 1), "pySensors value": schedule_matrix[0, selected], "pySensors x": X[space_idx], "pySensors y": Y[space_idx], "pySensors time": times[time_idx]})
raven_compare = pd.DataFrame({"column_id": raven_column_id, "sensor_id": raven_space_idx, "RAVEN order": raven_sensor_table["sensor"].to_numpy(), "RAVEN value": raven_sensor_table["T"].to_numpy(), "RAVEN x": raven_sensor_table["x"].to_numpy(), "RAVEN y": raven_sensor_table["y"].to_numpy(), "RAVEN time": raven_sensor_table["time"].to_numpy(), "nearest_grid_distance": raven_distances})
comparison_table = pysensors_compare.merge(raven_compare, on=["column_id", "sensor_id"], how="outer").sort_values(["pySensors order", "RAVEN order"], na_position="last").reset_index(drop=True)
comparison_table["same sensor"] = comparison_table["pySensors order"].notna() & comparison_table["RAVEN order"].notna()
comparison_table["same location"] = np.isclose(comparison_table["pySensors x"], comparison_table["RAVEN x"]) & np.isclose(comparison_table["pySensors y"], comparison_table["RAVEN y"]) & np.isclose(comparison_table["pySensors time"], comparison_table["RAVEN time"])
comparison_table["same value"] = np.isclose(comparison_table["pySensors value"], comparison_table["RAVEN value"])
comparison_table["match"] = comparison_table[["same sensor", "same location", "same value"]].all(axis=1)
comparison_table["order_match"] = comparison_table["pySensors order"].eq(comparison_table["RAVEN order"])
print(f"same selected schedule set: {comparison_table['match'].all() and len(comparison_table) == n_sensors}")
print(f"same selected schedule order: {np.array_equal(selected, raven_column_id)}")
style_sensor_comparison(comparison_table, "RAVEN vs pySensors spatiotemporal schedule", columns=["match", "order_match", "column_id", "sensor_id", "pySensors order", "RAVEN order", "pySensors value", "RAVEN value", "pySensors x", "RAVEN x", "pySensors y", "RAVEN y", "pySensors time", "RAVEN time"])
```

## Schedule footprint and held-out reconstruction

```python
alpha_index = 2
frame = 3
fig, ax = plt.subplots(figsize=(6, 4.6))
sc = spatial_frame(ax, T[alpha_index, frame], f"Schedule footprint at time={times[frame]:.3f}", (T.min(), T.max()), sensors=space_idx, schedule_times=times[time_idx], frame_time=times[frame])
ax.set_ylabel("y")
fig.colorbar(sc, ax=ax, label="temperature")
plt.show()
```

```python
heldout_alpha_index = 2
train_schedule = np.delete(schedule_matrix, heldout_alpha_index, axis=0)
test_schedule = schedule_matrix[heldout_alpha_index][None, :]
heldout_model = ps.SSPOR(basis=ps.basis.Identity(n_basis_modes=train_schedule.shape[0]), n_sensors=n_sensors, optimizer=ps.optimizers.QR())
heldout_model.fit(train_schedule, seed=42)
heldout_selected = heldout_model.get_selected_sensors()
heldout_space_idx = heldout_selected // len(times)
heldout_time_idx = heldout_selected % len(times)
regularized_reconstruction = heldout_model.predict(test_schedule[:, heldout_selected])[0]
test_truth = test_schedule[0]
regularized_error = np.linalg.norm(test_truth - regularized_reconstruction) / np.linalg.norm(test_truth)
test_reconstruction = heldout_model.predict(test_schedule[:, heldout_selected], method="unregularized")[0]
test_error = np.linalg.norm(test_truth - test_reconstruction) / np.linalg.norm(test_truth)
truth_grid = test_truth.reshape(X.size, len(times)).T
recon_grid = test_reconstruction.reshape(X.size, len(times)).T
error_grid = recon_grid - truth_grid
relative_error_grid = error_grid / max(np.max(np.abs(truth_grid)), 1e-12)
time_relative_l2_error = np.linalg.norm(error_grid, axis=1) / np.maximum(np.linalg.norm(truth_grid, axis=1), 1e-12)
print(f"held-out alpha: {alphas[heldout_alpha_index]:.3f}")
print(f"regularized default error: {regularized_error:.3e}")
print(f"unregularized held-out spatiotemporal error: {test_error:.3e}")
pd.DataFrame({"order": np.arange(1, len(heldout_selected) + 1), "column_id": heldout_selected, "point_id": heldout_space_idx, "x": X[heldout_space_idx], "y": Y[heldout_space_idx], "time": times[heldout_time_idx]})
```

```python
field_limits = (truth_grid.min(), truth_grid.max())
error_limit = max(np.max(np.abs(relative_error_grid)), 1e-12)
fig, axes = plt.subplots(1, 3, figsize=(13.5, 4.2), sharex=True, sharey=True)
truth_sc = spatial_frame(axes[0], truth_grid[0], "Truth", field_limits, sensors=heldout_space_idx, schedule_times=times[heldout_time_idx], frame_time=times[0])
recon_sc = spatial_frame(axes[1], recon_grid[0], "Reconstruction", field_limits, sensors=heldout_space_idx, schedule_times=times[heldout_time_idx], frame_time=times[0])
err_sc = spatial_frame(axes[2], relative_error_grid[0], "Relative error", (-error_limit, error_limit), cmap="seismic", sensors=heldout_space_idx, schedule_times=times[heldout_time_idx], frame_time=times[0])
for ax in axes:
    ax.set_ylabel("y")
fig.colorbar(truth_sc, ax=axes[0], shrink=0.8, label="temperature")
fig.colorbar(recon_sc, ax=axes[1], shrink=0.8, label="temperature")
fig.colorbar(err_sc, ax=axes[2], shrink=0.8, label="relative error")

def update(frame):
    truth_sc.set_data(field_image(truth_grid[frame]))
    recon_sc.set_data(field_image(recon_grid[frame]))
    err_sc.set_data(field_image(relative_error_grid[frame]))
    fig.suptitle(f"held-out alpha={alphas[heldout_alpha_index]:.3f}, time={times[frame]:.3f}; rel L2={time_relative_l2_error[frame]:.3e}")
    return truth_sc, recon_sc, err_sc

spatiotemporal_field_animation = animation.FuncAnimation(fig, update, frames=len(times), interval=650, blit=False)
plt.close(fig)
HTML(spatiotemporal_field_animation.to_jshtml())
```
